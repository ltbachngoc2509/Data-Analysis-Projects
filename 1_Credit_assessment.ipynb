```python
#2.8 Xử lý dữ liệu mất cân bằng
X = df.drop( columns=['NHOMNO', 'NHOMNOMOI', 'NHOMNO_TCBS'] )

y = df['NHOMNO']
```

```python
y.value_counts()
```

```python
from imblearn.over_sampling import SMOTE
import numpy as np
import pandas as pd

# Ensure columns are datetime type in X, coercing any errors to NaT
X['OPEN_DATE'] = pd.to_datetime(X['OPEN_DATE'], dayfirst=True, errors='coerce')
X['NGAYDENHAN'] = pd.to_datetime(X['NGAYDENHAN'], dayfirst=True, errors='coerce')

# Convert datetime objects to Unix timestamp (seconds), converting NaT to NaN
X['OPEN_DATE'] = X['OPEN_DATE'].apply(lambda x: x.timestamp() if pd.notna(x) else np.nan)
X['NGAYDENHAN'] = X['NGAYDENHAN'].apply(lambda x: x.timestamp() if pd.notna(x) else np.nan)

# Impute NaN values with the median, as SMOTE cannot handle NaNs
for col in ['OPEN_DATE', 'NGAYDENHAN']:
    if X[col].isna().any():
        median_val = X[col].median()
        X[col].fillna(median_val, inplace=True)

smote = SMOTE(random_state=42)

X_resampled, y_resampled = smote.fit_resample(X,y)
```

```python
print("Trước SMOTE:")
print(y.value_counts())

print("\nSau SMOTE:")
print(y_resampled.value_counts())
```

```python
#2.9 Chuẩn hóa dữ liệu
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X_resampled)
```

```python
#2.10 Chia tập dữ liệu train/
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y_resampled,
    test_size=0.2,
    random_state=42,
    stratify=y_resampled
)
```

```python
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
#3.1 Mo hình truyền thống( logistic regression) - chưa tinh chỉnh tham số
#LR sẽ cố gắng tìm một công thức toán học để phân biệt khách hàng tốt và xấu
#Huấn luyện mô hình
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(
    random_state=42,
    max_iter=500
)

lr.fit(X_train, y_train)
```

Hàm fit() được sử dụng để huấn luyện mô hình trên tập dữ liệu huấn luyện. Trong quá trình này, mô hình học mối quan hệ giữa 19 biến đầu vào và biến mục tiêu NHOMNO, từ đó xây dựng hàm dự báo phục vụ cho việc phân loại khách hàng theo nhóm nợ.

Sử dụng mô hình Logistic Regression để thực hiện bài toán phân loại nhóm nợ khách hàng. Mô hình được khởi tạo với tham số random_state = 42 nhằm đảm bảo kết quả có thể tái lặp giữa các lần chạy và max_iter = 500 để tăng số vòng lặp tối đa, giúp thuật toán hội tụ tốt hơn. Sau đó, phương thức fit(X_train, y_train) được sử dụng để huấn luyện mô hình trên tập dữ liệu huấn luyện gồm 85.296 quan sát và 19 biến đầu vào. Trong quá trình huấn luyện, mô hình học mối quan hệ giữa các biến độc lập như loại khách hàng, giới tính, dư nợ, lãi suất, thời gian vay,… với biến mục tiêu NHOMNO, từ đó xây dựng hàm dự báo xác suất khách hàng thuộc từng nhóm nợ.

```python
y_pred_lr = lr.predict(X_test)
y_prob_lr = lr.predict_proba(X_test)
```

Sau khi hoàn thành quá trình huấn luyện, mô hình được sử dụng để dự đoán trên tập kiểm tra. Hàm predict() trả về nhãn dự đoán của từng quan sát, trong khi predict_proba() trả về xác suất khách hàng thuộc lớp mục tiêu. Các kết quả dự đoán này sẽ được sử dụng ở phần sau để tính toán các chỉ số đánh giá như Accuracy, Precision, Recall, F1-score, ROC-AUC và xây dựng ma trận nhầm lẫn (Confusion Matrix)

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV

lr = LogisticRegression(random_state=42)

param_grid_lr = {
    'C': [0.001, 0.01, 0.1, 1, 10],
    'penalty': ['l2'],
    'solver': ['lbfgs'],
    'max_iter': [200, 500, 1000]
}

grid_lr = GridSearchCV(
    estimator=lr,
    param_grid=param_grid_lr,
    cv=5,
    scoring='accuracy',   # hoặc 'roc_auc_ovr' nếu đa lớp
    n_jobs=-1
)

grid_lr.fit(X_train, y_train)

print("Best params:", grid_lr.best_params_)
print("Best score:", grid_lr.best_score_)
```

Do kích thước dữ liệu sau khi cân bằng bằng SMOTE khá lớn, việc sử dụng GridSearchCV với nhiều tổ hợp siêu tham số làm tăng đáng kể thời gian huấn luyện. Vì vậy, nghiên cứu lựa chọn phương pháp tinh chỉnh thủ công (manual hyperparameter tuning) trên một số tổ hợp tham số quan trọng (n_estimators, max_depth) để cân bằng giữa hiệu quả mô hình và chi phí tính toán.

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

configs = [
    {"n_estimators":100, "max_depth":None},
    {"n_estimators":200, "max_depth":None},
    {"n_estimators":300, "max_depth":None},
    {"n_estimators":100, "max_depth":15},
    {"n_estimators":200, "max_depth":15},
    {"n_estimators":300, "max_depth":20},
]

best_acc = 0
best_model = None
best_params = None

for params in configs:

    rf = RandomForestClassifier(
        n_estimators=params["n_estimators"],
        max_depth=params["max_depth"],
        random_state=42,
        n_jobs=-1
    )

    rf.fit(X_train, y_train)

    y_pred = rf.predict(X_test)

    acc = accuracy_score(y_test, y_pred)

    print(params, acc)

    if acc > best_acc:
        best_acc = acc
        best_model = rf
        best_params = params

print(best_params)
print(best_acc)
```

```python
import pandas as pd

# Make predictions with the best Logistic Regression model found by GridSearchCV
y_pred_best_lr = grid_lr.best_estimator_.predict(X_test)
y_prob_best_lr = grid_lr.best_estimator_.predict_proba(X_test)

# Make predictions with the best Random Forest model found during manual tuning
y_pred_best_rf = best_model.predict(X_test)
y_prob_best_rf = best_model.predict_proba(X_test)

results = pd.DataFrame({
    "Model": [
        "LR Before",
        "LR After",
        "RF Before",
        "RF After"
    ],
    "Accuracy": [
        accuracy_score(y_test, y_pred_lr),
        accuracy_score(y_test, y_pred_best_lr),
        accuracy_score(y_test, y_pred_rf),
        accuracy_score(y_test, y_pred_best_rf)
    ],
    "Precision": [
        precision_score(y_test, y_pred_lr, average='weighted'),
        precision_score(y_test, y_pred_best_lr, average='weighted'),
        precision_score(y_test, y_pred_rf, average='weighted'),
        precision_score(y_test, y_pred_best_rf, average='weighted')
    ],
    "Recall": [
        recall_score(y_test, y_pred_lr, average='weighted'),
        recall_score(y_test, y_pred_best_lr, average='weighted'),
        recall_score(y_test, y_pred_rf, average='weighted'),
        recall_score(y_test, y_pred_best_rf, average='weighted')
    ],
    "F1-score": [
        f1_score(y_test, y_pred_lr, average='weighted'),
        f1_score(y_test, y_pred_best_lr, average='weighted'),
        f1_score(y_test, y_pred_rf, average='weighted'),
        f1_score(y_test, y_pred_best_rf, average='weighted')
    ],
    "AUC": [
        roc_auc_score(y_test, y_prob_lr, multi_class='ovr', average='weighted'),
        roc_auc_score(y_test, y_prob_best_lr, multi_class='ovr', average='weighted'),
        roc_auc_score(y_test, y_prob_rf, multi_class='ovr', average='weighted'),
        roc_auc_score(y_test, y_prob_best_rf, multi_class='ovr', average='weighted')
    ]
})

results
```

Random Forest: Sau khi tinh chỉnh siêu tham số, mô hình có sự cải thiện nhẹ ở tất cả các chỉ số đánh giá.
```python
from sklearn.preprocessing import LabelEncoder

# --- Step 1: Convert numeric columns that were incorrectly loaded as objects to numeric ---
# Convert 'BASE_BAL', 'DUNO_QD', 'LAISUAT' to numeric, handling commas and errors
for col_to_convert in ['BASE_BAL', 'DUNO_QD', 'LAISUAT']:
    # Replace comma with dot for decimal conversion
    if oos[col_to_convert].dtype == 'object':
        oos[col_to_convert] = oos[col_to_convert].astype(str).str.replace(',', '.', regex=False)
    oos[col_to_convert] = pd.to_numeric(oos[col_to_convert], errors='coerce')

    # Fill NaN values with the median from the original df
    if oos[col_to_convert].isnull().any():
        median_val = df[col_to_convert].median() # Use median from original df
        oos[col_to_convert].fillna(median_val, inplace=True)

# --- Step 2: Drop 'CURR_BAL' as it was dropped from df ---
if 'CURR_BAL' in oos.columns:
    oos = oos.drop(columns=['CURR_BAL'])

# --- Step 3: Apply LabelEncoder to actual object columns ---
cat_cols_oos = oos.select_dtypes(include='object').columns

for col in cat_cols_oos:
    le = label_encoders[col] # Get the fitted LabelEncoder for the column

    # Identify unseen labels in oos that are not in the encoder's classes
    unseen_labels = set(oos[col].astype(str).unique()) - set(le.classes_)

    if unseen_labels: # If there are unseen labels
        # Get the mode of the original (encoded) df column, then inverse transform to get the string
        # This gives us a valid string label that the encoder knows
        # Need to ensure df[col] is the original, un-encoded column before getting mode
        # Assuming df[col] in `label_encoders[col].inverse_transform` refers to its pre-encoded form
        # The `df` used here is the one where LabelEncoder was fitted.

        # A safer approach for handling unseen labels is to map them to a specific known category
        # or the most frequent category seen during training.
        # Let's use the most frequent category from the training data for that column.
        # The `label_encoders` dict stores the fitted encoder, so `le.classes_` contains all known labels.

        # Find the most frequent label in the original training data for this column
        # We need the original df values before encoding to find the mode in terms of labels
        # However, the df has already been encoded. We can use le.inverse_transform to get the labels.

        # Re-derive original string values from df for mode calculation if necessary, or
        # use a default 'unknown' or the most frequent known label.
        # Simplest is to map to the most frequent label that the encoder knows.

        # Get the inverse transform of the mode of the *encoded* df column
        # This will give us the most frequent *string* label from the original df
        most_frequent_known_label_str = le.inverse_transform([df[col].mode()[0]])[0]

        # Replace unseen labels in oos with the most frequent known label
        oos[col] = oos[col].astype(str).replace(list(unseen_labels), most_frequent_known_label_str)

    # Now, transform the column (all labels should be known)
    oos[col] = le.transform(oos[col].astype(str))
```
