```python
#4.4 Kiểm tra dữ liệu
# Vẽ boxplot cho 6 đặc trưng đầu tiên để kiểm tra outliers
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(2, 3, figsize=(15, 10))
features = df.columns[:6]  # Lấy 6 cột đầu

for i, feature in enumerate(features):
    row = i // 3
    col = i % 3
    sns.boxplot(x='target_label', y=feature, data=df, ax=axes[row, col])
    axes[row, col].set_title(f'Boxplot của {feature}')
    axes[row, col].set_xlabel('')

plt.tight_layout()
plt.show()
```

```python
#4.4 Kiểm tra dữ liệu ảo, sai

def count_outliers(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return ((data[column] < lower) | (data[column] > upper)).sum()
print("Số lượng outliers trong từng đặc trưng:")
for feature in features:
    n = count_outliers(df, feature)
    print(f"{feature}: {n} outliers")
```

```python
#4.5 CHUẨN HÓA DỮ LIỆU
from sklearn.preprocessing import StandardScaler

# Lấy các cột đặc trưng
feature_columns = df.drop(['target', 'target_label'], axis=1).columns

# Khởi tạo scaler
scaler = StandardScaler()

# Thực hiện scale
df_scaled = df.copy()
df_scaled[feature_columns] = scaler.fit_transform(df[feature_columns])
print(f"Trước scale: mean={df['mean radius'].mean():.2f}, std={df['mean radius'].std():.2f}")
print(f"Sau scale  : mean={df_scaled['mean radius'].mean():.2f}, std={df_scaled['mean radius'].std():.2f}")
```

```python
#4.6 Kiểm tra tính tương quan của dữ liệu
import seaborn as sns
import matplotlib.pyplot as plt

# Lấy các cột đặc trưng
feature_columns = df.drop(['target', 'target_label'], axis=1).columns

# Tính ma trận tương quan
corr_matrix = df[feature_columns].corr()

# Vẽ heatmap
plt.figure(figsize=(20, 16))
sns.heatmap(corr_matrix, annot=False, cmap='coolwarm', center=0)
plt.title('Ma trận tương quan giữa các đặc trưng')
plt.show()
```

```python
# Tìm các cặp có tương quan > 0.9
threshold = 0.9
high_corr_pairs = []

for i in range(len(corr_matrix.columns)):
    for j in range(i+1, len(corr_matrix.columns)):
        if abs(corr_matrix.iloc[i, j]) > threshold:
            high_corr_pairs.append((corr_matrix.columns[i],
                                   corr_matrix.columns[j],
                                   corr_matrix.iloc[i, j]))

print(f"Các cặp đặc trưng có tương quan > {threshold}:")
for pair in high_corr_pairs[:10]:  # In 10 cặp đầu
    print(f"{pair[0]} - {pair[1]}: {pair[2]:.3f}")

print(f"\nTổng số cặp tương quan cao: {len(high_corr_pairs)}")
```

```python
#4.7 TÁCH DỮ LIỆU
from sklearn.model_selection import train_test_split

# Lấy features và target
X = df.iloc[:, :-2]  # Lấy tất cả cột trừ 2 cột cuối (target và target_label)
y = df['target']      # Cột target

# Lần 1: Tách 70% train, 30% còn lại (sẽ chia thành val và test)
X_train, X_tmp, y_train, y_tmp = train_test_split(
    X, y,
    test_size=0.3,
    random_state=42
)

# Lần 2: Từ 30% còn lại, tách 50% val, 50% test
X_val, X_test, y_val, y_test = train_test_split(
    X_tmp, y_tmp,
    test_size=0.5,
    random_state=42
)

# Xem kết quả
print("Kích thước các tập:")
print("Train:", X_train.shape)
print("Validation:", X_val.shape)
print("Test:", X_test.shape)
```

## 5. MÔ HÌNH HÓA DỮ LIỆU

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(X_test)

#HUẤN LUYỆN MÔ HÌNH:
from sklearn.linear_model import LogisticRegression

lr_model = LogisticRegression(max_iter=1000, random_state=42)
lr_model.fit(X_train_scaled, y_train)
#DỰ ĐOÁN VÀ ĐÁNH GIÁ:
from sklearn.metrics import accuracy_score, classification_report

y_pred = lr_model.predict(X_val_scaled)

print("Accuracy:", accuracy_score(y_val, y_pred))
print(classification_report(y_val, y_pred))
```

```python
#RANDOM FOREST
# Import thư viện
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# Tạo mô hình Random Forest
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
# Huấn luyện mô hình (dùng dữ liệu gốc, KHÔNG scale)
rf_model.fit(X_train, y_train)

# Dự đoán trên tập validation
y_pred_rf = rf_model.predict(X_val)

# Đánh giá kết quả
print(" KẾT QUẢ RANDOM FOREST ")
print(f"Accuracy: {accuracy_score(y_val, y_pred_rf):.4f}")
print("\nBáo cáo phân loại:")
print(classification_report(y_val, y_pred_rf, target_names=['Malignant (0)', 'Benign (1)']))

# Ma trận nhầm lẫn
print("\nConfusion Matrix:")
print(confusion_matrix(y_val, y_pred_rf))
# Xem độ quan trọng của từng đặc trưng (feature importance)
# Lấy top 5 đặc trưng quan trọng nhất
import numpy as np
feature_importance = pd.DataFrame({
    'Đặc trưng': feature_columns,
    'Độ quan trọng': rf_model.feature_importances_
}).sort_values('Độ quan trọng', ascending=False)

print("\nTop 5 đặc trưng quan trọng nhất:")
print(feature_importance.head())
```

```python
#GRADIENT BOOSTING
# Import thư viện
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Tạo mô hình
gb_model = GradientBoostingClassifier(
    n_estimators=100,
    learning_rate=0.1,
    random_state=42
)

# Huấn luyện
gb_model.fit(X_train, y_train)

# Dự đoán
y_pred_gb = gb_model.predict(X_val)

# Đánh giá
print("=== KẾT QUẢ GRADIENT BOOSTING ===")
print(f"Accuracy: {accuracy_score(y_val, y_pred_gb):.4f}")

print("\nBáo cáo phân loại:")
print(classification_report(y_val, y_pred_gb,
                            target_names=['Malignant (0)', 'Benign (1)']))

# Confusion matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_val, y_pred_gb))
```

```python
#5.2 ĐÁNH GIÁ BẰNG CÁC THANG ĐO PHÙ HỢP
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Dự đoán trên tập test
y_pred_lr = lr_model.predict(X_test_scaled)
y_pred_rf = rf_model.predict(X_test)
y_pred_gb = gb_model.predict(X_test)

# Tính chỉ số
acc_lr = accuracy_score(y_test, y_pred_lr)
pre_lr = precision_score(y_test, y_pred_lr)
rec_lr = recall_score(y_test, y_pred_lr)
f1_lr = f1_score(y_test, y_pred_lr)

acc_rf = accuracy_score(y_test, y_pred_rf)
pre_rf = precision_score(y_test, y_pred_rf)
rec_rf = recall_score(y_test, y_pred_rf)
f1_rf = f1_score(y_test, y_pred_rf)

acc_gb = accuracy_score(y_test, y_pred_gb)
pre_gb = precision_score(y_test, y_pred_gb)
rec_gb = recall_score(y_test, y_pred_gb)
f1_gb = f1_score(y_test, y_pred_gb)

# In bảng so sánh
print("\n SO SÁNH KẾT QUẢ CÁC MÔ HÌNH TRÊN TẬP TEST \n")
print(f"{'Mô hình':<20} {'Accuracy':<10} {'Precision':<10} {'Recall':<10} {'F1-Score':<10}")
print("-" * 60)
print(f"{'Logistic Regression':<20} {acc_lr:<10.4f} {pre_lr:<10.4f} {rec_lr:<10.4f} {f1_lr:<10.4f}")
print(f"{'Random Forest':<20} {acc_rf:<10.4f} {pre_rf:<10.4f} {rec_rf:<10.4f} {f1_rf:<10.4f}")
print(f"{'Gradient Boosting':<20} {acc_gb:<10.4f} {pre_gb:<10.4f} {rec_gb:<10.4f} {f1_gb:<10.4f}")
print("-" * 60)
```
